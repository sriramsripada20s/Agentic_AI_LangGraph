```python
from langgraph.graph import StateGraph, START, END
from langchain_anthropic import ChatAnthropic
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatAnthropic(model="claude-haiku-5-5")
```

```python
# create state

class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state:LLMState) -> LLMState:
    #extract the question from State
    question = state["question"]
    #form the prompt to send to the LLM
    prompt = f"Answer the following question: {question}"
    #ask that question to LLM and get the answer
    answer = model.invoke(prompt)
    #update the state with the answer and return it
    state['answer'] = answer
    return state
```

```python
#create our Graph
graph = StateGraph(LLMState)

#add nodes to our graph
graph.add_node('llm_qa',llm_qa)

#add_edges to our graph
graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)

workflow = graph.compile()
```

```python
# execute

intial_state = {'question': 'How far is moon from the earth?'}

final_state = workflow.invoke(intial_state)

print(final_state['answer'])
```
